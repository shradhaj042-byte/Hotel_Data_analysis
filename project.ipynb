### Data Analysis and Visualizations

```python
cancelled_perc = df['is_canceled'].value_counts(normalize= True)
print(cancelled_perc)

plt.Figure(figsize= (5,4))
plt.title('Reservation Status count')
plt.bar(['Not canceled', 'Canceled'],df['is_canceled'].value_counts(),edgecolor = 'k', width =0.7)
plt.show()
```


#### Reservation Cancellation Analysis

- Total cancelled reservations: **44,224**
- Total non-cancelled reservations: **75,166**
- Around 37% of hotel reservations were cancelled.
- This analysis helps understand the hotel's cancellation pattern.


```python
plt.figure(figsize=(8,5))

ax = sns.countplot(x='hotel', hue='is_canceled', data=df, palette='Blues')

ax.legend(['Not Cancelled', 'Cancelled'], loc='upper left')

plt.title('Hotel-wise Cancellation Analysis')
plt.xlabel('Hotel')
plt.ylabel(' Number of Bookings')

plt.show()
```


#### Hotel-wise Cancellation Analysis

- City Hotel recorded **33,102 cancelled bookings**.
- Resort Hotel recorded **11,122 cancelled bookings**.
- City Hotel had approximately **75%** of all cancelled bookings.
- Resort Hotel had approximately **25%** of all cancelled bookings.
- This indicates that City Hotel experienced significantly more booking cancellations than Resort Hotel.

-Total Cancelled Bookings: 44,224**


```python
resort_hotel = df[df['hotel'] == 'Resort Hotel']

resort_hotel['is_canceled'].value_counts(normalize=True)
```

```python
city_hotel = df[df['hotel'] == 'City Hotel']

city_hotel['is_canceled'].value_counts(normalize = True)
```

```python
resort_hotel = resort_hotel.groupby('reservation_status_date')[['adr']].mean()
city_hotel = city_hotel.groupby('reservation_status_date')[['adr']].mean()
```

```python
plt.figure(figsize= (20,8))
plt.title('Average Daily Rate in City and Resort Hotel',fontsize = 30)
plt.plot(resort_hotel.index,resort_hotel ['adr'], label = 'Resort_Hotel')
plt.plot(city_hotel.index, city_hotel['adr'] , label= 'city_hotel')
plt.legend(fontsize = 20)
plt.show()
```


#### Average Daily Rate (ADR) Analysis

- This line chart compares the Average Daily Rate (ADR) of City Hotel and Resort Hotel.
- It shows how hotel room prices change over time.
- The chart helps identify differences in pricing trends between the two hotels.
- It also helps understand daily fluctuations in room rates.


```python
df['month'] = df['reservation_status_date'].dt.month
plt.figure(figsize = (16,8))
ax1 = sns.countplot(x= 'month',hue = 'is_canceled', data = df , palette= 'bright')
legend_lables,_ = ax1.get_legend_handles_labels()
ax1.legend(bbox_to_anchor=(1,1))
plt.title('Reservation Status Per Month',size = 20)
plt.xlabel('month')
plt.ylabel('number of reservations')
plt.legend(['not canceled', 'canceled'])
plt.show
```


#### Monthly Reservation Status Analysis

- January recorded the highest number of cancelled reservations (6,061).
- September recorded the lowest number of cancelled reservations (2,676).
- August had the highest number of non-cancelled reservations (8,561).
- December had the lowest number of non-cancelled reservations (3,873)

January experienced the highest cancellations, while September had the lowest cancellations. Hotels should focus on reducing cancellations during months with higher cancellation rates.
This analysis is based on the month extracted from `reservation_status_date`.


```python
plt.figure(figsize=(15,8))
plt.title('ADR Per Month', fontsize=30)

sns.barplot(
    x='month',
    y='adr',
    data=df[df['is_canceled'] == 1].groupby('month')[['adr']].sum().reset_index(),
    palette='bright',
    hue='month',
    legend=False
)

plt.show()
```


#### Monthly ADR Analysis of Cancelled Reservations

- This chart shows the total ADR of cancelled reservations for each month.
- It helps identify which months recorded the highest and lowest ADR values.
- The monthly ADR comparison helps understand revenue patterns among cancelled bookings.




```python
cancelled_data = df[df['is_canceled'] == 1]
top_10_country = cancelled_data['country'].value_counts()[:10]
plt.figure(figsize = (8,7))
plt.title('Top 10 countries with reservation canceled')
plt.pie(top_10_country, autopct= '%.2f', labels = top_10_country.index)
plt.show()
```

#### Top 10 Countries with Cancelled Reservations

- This pie chart shows the top 10 countries with the highest number of cancelled reservations.
- It helps identify which countries contribute the most to booking cancellations.
- Portugal recorded the highest number of cancelled reservations.
- The chart highlights the major countries contributing to hotel booking cancellations.

Hotels can focus on understanding customer booking patterns from countries with higher cancellation counts and develop better booking strategies to reduce cancellations.
