#### Business Problem 

The hotel loses money when guests cancel. A cancelled room may stay empty, and the hotel has already planned staff, food and cleaning for that guest.

Main question: Which bookings are most likely to be cancelled, how much money does this cost us, and what can we do to reduce it?

An end-to-end exploratory data analysis of hotel booking and cancellation data.

### Data Analysis and Visualizations

```python
cancelled_perc = df['is_canceled'].value_counts(normalize= True)
print(cancelled_perc)

plt.Figure(figsize= (5,4))
plt.title('Reservation Status count')
plt.bar(['Not canceled', 'Canceled'],df['is_canceled'].value_counts(),edgecolor = 'k', width =0.7)
plt.show()
```
